```python
"""
============================================================================
Author          : Mohammad Norizadeh Cherloo
Website         : https://onlinebme.com/
GitHub          : https://github.com/Mohammad-Norizadeh-Cherloo

Project         : Visualization of Optimization Algorithms
Algorithm       : AdaGrad

Description     : Visualizes the AdaGrad optimizer. It adapts the learning 
                  rate for each parameter by accumulating the squared 
                  gradients, giving larger updates to infrequent features.
============================================================================
"""
import numpy as np
from matplotlib import pyplot as plt
```

```python
# cost= lambda w: w**4
# grad= lambda w:4*w**3

cost= lambda w: 0.3*w**4-0.1*w**3-2*w**2-0.8*w
grad= lambda w:1.2*w**3- 0.3*w**2- 4*w-0.8
```

```python
w0=np.array([-2.8])
j0= cost(w0)
```

SGD 

```python

lr=0.1
w=w0
epoch=1000

wn=[]
jn=[]
gn=[]
for iter in range(1,epoch+1):
     wn.append(w)
     jn.append(cost(w))
     
     g= grad(w)
     w= w-lr*g
     
     gn.append(g)

wn= np.array(wn)
jn= np.array(jn)
gn= np.array(gn)

x=np.arange(-3,3,0.1)
y= cost(x)
plt.figure(figsize=(12, 6))
plt.plot(x,y,linewidth=3)
plt.plot(w0+0.05,j0,'ro')
plt.plot(wn+0.05,jn,'-',linewidth=2)
color= np.arange(wn.size)
plt.scatter(wn+0.05,jn,s=np.abs(gn)*50,c=color)
plt.plot(wn[-1]+0.05,jn[-1],'bo')
plt.show()
```

AdaGrad (Adaptive gradients)

```python
lr=0.99
Gt=0
eps= 10e-8
w1=w0
wn1=[]
jn1=[]
gn1=[]
for t in range(1,epoch+1):
     wn1.append(w1)
     jn1.append(cost(w1))
     
     gt= grad(w1)
     Gt= Gt+ gt*gt
     dw= -(lr/np.sqrt(Gt+eps))*gt
     w1= w1 + dw
     
     gn1.append(gt)

wn1= np.array(wn1)
jn1= np.array(jn1)
gn1= np.array(gn1)

x=np.arange(-3,3,0.1)
y= cost(x)
plt.figure(figsize=(12, 6))
plt.plot(x,y,linewidth=3)
plt.plot(w0+0.05,j0,'ro')
plt.plot(wn1+0.05,jn1,'-',linewidth=2)
color= np.arange(wn1.size)
plt.scatter(wn1+0.05,jn1,s=np.abs(gn1)*50,c=color)
plt.plot(wn1[-1]+0.05,jn1[-1],'bo')
plt.show()
```

```python
plt.figure(figsize=(12, 6))
plt.plot(gn,label='SGD',linewidth=3)
plt.plot(gn1,label='SGD+Momentum',linewidth=3)
plt.legend()
plt.show()
```
